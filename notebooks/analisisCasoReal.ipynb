# Caso real de Big Data: Netflix
## 1. ¿Qué datos genera?

Netflix genera enormes cantidades de datos cada segundo debido a los millones de usuarios que utilizan la plataforma en todo el mundo. Entre los datos que produce se encuentran:

- Películas y series vistas por cada usuario.
- Tiempo que una persona pasa viendo contenido.
- Horarios en los que más se utiliza la plataforma.
- Dispositivos desde los que se accede (teléfono, TV, computadora, tablet).
- Búsquedas realizadas por los usuarios.
- Calificaciones y preferencias de contenido.
- Pausas, avances rápidos o abandono de episodios.

El volumen de datos es gigantesco, ya que millones de personas usan la plataforma diariamente. Los datos se generan en tiempo real y de manera continua las 24 horas del día.

## 2. ¿Quién los usa?

Los datos son utilizados por diferentes actores:

Netflix: analiza la información para mejorar su plataforma y aumentar la satisfacción de los usuarios.
Equipos de científicos de datos e ingenieros: desarrollan algoritmos de recomendación y análisis de comportamiento.
Productores y directores: utilizan estadísticas para saber qué tipo de contenido tiene más éxito.
Equipos de marketing: analizan preferencias para crear campañas más efectivas.

## 3. ¿Qué problema resuelve?

El análisis de Big Data permite a Netflix resolver varios problemas importantes:

Personalizar las recomendaciones para cada usuario.
Mejorar la experiencia del cliente mostrando contenido relevante.
Reducir costos al identificar qué series o películas son más exitosas.
Mantener a los usuarios más tiempo en la plataforma.
Predecir tendencias y decidir qué nuevas producciones realizar.

Gracias al análisis de datos, la empresa puede ofrecer contenido adaptado a los gustos de cada persona, aumentando la satisfacción y fidelidad de los clientes.

## 4. ¿Por qué no bastaría Excel?

Excel no sería suficiente por varias razones relacionadas con las características del Big Data:

### Volumen 
La cantidad de datos generados por millones de usuarios supera la capacidad de una hoja de cálculo tradicional.
### Velocidad 
Los datos se producen en tiempo real y necesitan procesarse rápidamente.
### Variedad 
Existen diferentes tipos de datos, como texto, videos, búsquedas y registros de comportamiento.
### Complejidad 
Se requieren algoritmos avanzados de inteligencia artificial y aprendizaje automático para analizar patrones y hacer recomendaciones.

Por estas razones, Netflix utiliza tecnologías especializadas de Big Data y computación en la nube en lugar de herramientas simples como Excel.
